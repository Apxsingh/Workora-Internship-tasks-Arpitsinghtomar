### Check for invalid values

```python
# Age
print("Invalid Age values :")
print(df[(df['Age'] < 18) | (df['Age'] > 100)])
```

```python
# Tenure

print("Invalid Tenures Values : ")
print(df[df['TenureMonths'] < 0])
```

```python
# Monthly Charges

print("Invalid Monthly Charges : ")
print(df[df['MonthlyCharges'] < 0])
```

```python
# Total Charges

print("Invalid Total Charges : ")
print(df[df['TotalCharges'] < 0])
```

```python
# SupportTickets

print("Invalid SuportTickits : ")
print(df[df['SupportTickets'] < 0])
```

### Validate Categorical Values

```python
# Gender Values
print("Gender Values : ")
print(df['Gender'].unique())


# Subscription Types
print("\nSubscription Types : ")
print(df['SubscriptionType'].unique())


# Contract Types
print("\nContract Types : ")
print(df['ContractType'].unique())


# Payment Methods
print("\nPayment Methods : ")
print(df['PaymentMethod'].unique())


# Churn Values
print("\nChurn Values : ")
print(df['Churn'].unique())
```

### Standardize Column Names

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(' ', '_')
)

print(df.columns)
```

### Standardize Categorical Strings

```python
categorical_columns = [
    'gender',
    'subscriptiontype',
    'contracttype',
    'paymentmethod',
    'churn'
]

for col in categorical_columns:
    df[col] = df[col].astype(str).str.strip()
```

### Validate Customer IDs

```python
print("Unique Customer IDs : ", df['customerid'].nunique())
print("Total rows : ", len(df))
```

```python
# Check Duplicates Customer IDs
print("Duplicate Customer IDs:", df['customerid'].duplicated().sum())
```

### Check the Relationship Between Charges

```python
df['expected_total'] = df['tenuremonths'] * df['monthlycharges']

df['charge_difference'] = (
    df['totalcharges'] - df['expected_total']
).abs()

print(df[['customerid',
          'totalcharges',
          'expected_total',
          'charge_difference']])
```

```python
print(
    "Rows with charge mismatch :",
    (df['charge_difference'] > 0.01).sum()
)
```

```python
df.drop(columns=['expected_total', 'charge_difference'], inplace=True)
```

### Final Validation

```python
print("Final Shape :",df.shape)
```

```python
# Missing Values
print("\nMissing Values :")
print(df.isnull().sum())


# Duplicate Rows
print("\nDuplicate Values :")
print(df.duplicated().sum())


# Data Types
print("\nData Types :")
print(df.dtypes)


# Final Dataset
print("\nFinal Dataset :")
print(df.head())
```
